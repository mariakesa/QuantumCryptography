```python
text = "attack at dawn"

# plaintext → bytes
pt = text.encode("ascii")

# ciphertext hex → bytes
ct = bytes.fromhex("6c73d5240a948c86981bc294814d")

# XOR plaintext and ciphertext to recover key
key = bytes(p ^ c for p, c in zip(pt, ct))

print(key)
print(key.hex())
pt

text_mal="attack at dusk"
malleable=text_mal.encode('ascii')
enc_mal = bytes(p ^ c for p, c in zip(malleable, key))
enc_mal.hex()
```
